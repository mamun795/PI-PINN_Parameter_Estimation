```python
import torch
import torch.nn as nn
import pandas as pd
import numpy as np
from scipy.integrate import odeint
```

```python
# Configuration parameters
beta = 0.5
gamma = 0.05
T = 100
num_samples = 120
S0 = 0.99
I0 = 0.01
R0 = 0.0
```

```python
# Define the SIR model ODEs
def sir_model(y, t, beta, gamma):
    S, I, R = y
    dS_dt = -beta * S * I
    dI_dt = beta * S * I - gamma * I
    dR_dt = gamma * I
    return dS_dt, dI_dt, dR_dt
```

```python
# Create synthetic data using the SIR model
def simulate_data(beta, gamma, T, num_samples, S0, I0, R0):
    t = np.linspace(0, T, num_samples).reshape(-1, 1)
    #print(t)
    y0 = [S0, I0, R0]
    params = (beta, gamma)
    y = odeint(sir_model, y0, t.flatten(), args=params)
    #print(y)
    S_data, I_data, R_data = y.T

    data_to_save = pd.DataFrame({'time': t.flatten(), 'Susceptible': S_data, 'Infected': I_data, 'Recovered': R_data})
    #print(data_to_save.head())
    # data_to_save.to_csv('sir_data.csv', index=False)
    data_to_save.to_csv('sir_data.csv', index=False)
    print(data_to_save)

    return t, S_data, I_data, R_data
```

```python
# Simulate data
t, S_data, I_data, R_data = simulate_data(beta, gamma, T, num_samples, S0, I0, R0)
```
